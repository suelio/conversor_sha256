Ferramenta para **pseudonimizacao e hashing criptografico** de dados tabulares (CSV, Excel, Parquet, JSON e TXT). Permite anonimizar colunas sensiveis especificas (PII como CPF, Email e Telefone), gerar impressoes digitais de registros com suporte a **Salt criptografico** e processamento em lote.

## 1. Importacao e Preparacao do Ambiente

```python
# Se estiver executando no Google Colab pela primeira vez:
# !git clone https://github.com/suelio/conversor_sha256.git
# %cd conversor_sha256

import pandas as pd
from hasher import DataHasher, HashAlgorithms, FileReader

print('Ambiente pronto e DataHasher importado com sucesso.')
```

## 2. Inspecao dos Dados Originais (Com Identificadores Pessoais)
Carregando a base de clientes com informacoes pessoais sensiveis antes do tratamento:

```python
df_original, meta = FileReader.read('data/clientes_exemplo.csv')
print('Metadados da base:', meta)
display(df_original)
```

## 3. Anonimizacao Seletiva de Colunas Sensiveis (CPF e Email)
Ao aplicar o hash com Salt (chave secreta), protegemos os dados contra ataques de forca bruta e Rainbow Tables:

```python
df_anonimizado = DataHasher.anonymize_dataframe(
    df=df_original,
    columns=['CPF', 'Email', 'Telefone'],
    algorithm='sha256',
    salt='chave_segura_lgpd_2026',
    keep_original=False,
    normalize=True
)

print('Base tratada (CPF, Email e Telefone anonimizados com SHA-256 + Salt):')
display(df_anonimizado)
```

## 4. Geracao de Impressao Digital (Fingerprint da Linha Inteira)
Cria um identificador unico criptografico baseado em todos os campos do registro:

```python
df_com_fingerprint = DataHasher.anonymize_dataframe(
    df=df_original,
    fingerprint_col='hash_identificador_registro',
    algorithm='sha256'
)

display(df_com_fingerprint[['ID_Cliente', 'Nome', 'hash_identificador_registro']])
```

## 5. Processamento e Exportacao Direta para Parquet / Excel
Executando o pipeline completo com metricas de tempo e velocidade de processamento:

```python
res = DataHasher.process_file(
    source='data/clientes_exemplo.csv',
    target_format='parquet',
    columns=['CPF', 'Email'],
    algorithm='sha256',
    salt='minha_chave_2026'
)

for chave, valor in res.items():
    print(f"{chave}: {valor}")
```
